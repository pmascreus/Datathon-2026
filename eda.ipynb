```python
import os
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import requests

# Carrega el .env tant si executes des de notebooks/ com des de l'arrel
env_path = Path("../.env") if Path("../.env").exists() else Path(".env")
for line in env_path.read_text().splitlines():
    if line.strip() and not line.strip().startswith("#"):
        key, _, value = line.partition("=")
        os.environ.setdefault(key.strip(), value.strip())

API = os.environ.get("RAWTREE_API_URL", "https://api.rawtree.com")
TEAM = os.environ["TEAM_ID"]

http = requests.Session()
http.headers["Authorization"] = f"Bearer {os.environ['RAWTREE_API_KEY']}"

def sql(q: str) -> pd.DataFrame:
    r = http.post(f"{API}/v1/query", json={"sql": q}, timeout=120)
    r.raise_for_status()
    return pd.DataFrame(r.json()["data"])

print(f"Connectat com a equip: {TEAM}")
```
